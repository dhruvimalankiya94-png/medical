# Diabetes Risk Prediction - Data Exploration & Feature Analysis
**Dataset**: Pima Indians Diabetes Database
**Objective**: Analyze physiological features and train a Random Forest Classifier for diabetes risk estimation.

```python
import pandas as pd
import numpy as np

# Load Dataset
df = pd.read_csv('../dataset/diabetes.csv')
print('Dataset Shape:', df.shape)
df.head()
```

```python
# Check for invalid zero values
zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
for col in zero_cols:
    zeros = (df[col] == 0).sum()
    print(f'Column {col}: {zeros} zero values ({zeros/len(df)*100:.1f}%)')
```
